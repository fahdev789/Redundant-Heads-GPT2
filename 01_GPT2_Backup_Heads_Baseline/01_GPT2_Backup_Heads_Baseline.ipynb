```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

import torch
from transformer_lens.model_bridge import TransformerBridge


SEED = 42

random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)

if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)

device = "cuda" if torch.cuda.is_available() else "cpu"

print("Device:", device)
print("PyTorch:", torch.__version__)
```

```python
# Load GPT-2 Medium
device = "cuda" if torch.cuda.is_available() else "cpu"

model = TransformerBridge.boot_transformers(
    "gpt2-medium",
    device=device
)

model.eval()


print("Layers:", model.cfg.n_layers)
print("Heads per layer:", model.cfg.n_heads)
print("Residual Dimension:", model.cfg.d_model)
print("Head dimension:", model.cfg.d_head)
```

```python
model.to_tokens("Patricia")
```

```python
name_pairs = [
    ("Alice", "Bob"),
    ("Mary", "John"),
    ("David", "Michael"),
    ("Sarah", "James"),
    ("Robert", "William"),
    ("Emma", "Daniel"),
    ("Linda", "Richard"),
    ("Jennifer", "Thomas"),
    ("Patricia", "Charles"),
    ("Barbara", "Christopher")
]

prompt = [
    f"When {a} and {b} went to the store, "
    f"{b} gave a drink to"
    for a, b in name_pairs
]


correct_names = [a for a, b in name_pairs]
incorrect_names = [b for a, b in name_pairs]

dataset = pd.DataFrame({
    "prompt": prompt,
    "correct_name": correct_names,
    "incorrect_name": incorrect_names
})

dataset.head()
```

```python
# Here the tokenization needs to be verified


tokens = model.to_tokens(prompt)

print("Token shape:", tokens.shape)

for prompt, token_row in zip(dataset["prompt"], tokens):
    print("\n", prompt)
    print(model.to_str_tokens(token_row))


assert len(set(tokens.shape[1:] for _ in prompt)) == 1, "All prompts must have the same length"
```

```python
# verify explicitly target name should be one token with its leading space

for name in set(correct_names + incorrect_names):
    token_id = model.to_single_token(" " + name)
    print(name, "->", token_id)
```

```python
# Calculate logit IOT difference per-example

# Make sure the baseline will perfect, if otherwise improve the prompt distribution before proceeding.

def get_ioi_scores(logits, correct_names, incorrect_names):
    """
    Calculate per-example IOI logit differences.
    """

    scores = []

    for i, (correct, incorrect) in enumerate(
        zip(correct_names, incorrect_names)
    ):

        correct_id = model.to_single_token(" " + correct)
        incorrect_id = model.to_single_token(" " + incorrect)

        correct_logit = logits[i, -1, correct_id]
        incorrect_logit = logits[i, -1, incorrect_id]

        # here we will use logit metric
        scores.append(
            (correct_logit - incorrect_logit).item()
        )

    return np.array(scores)


def evaluate_ioi(logits):
    scores = get_ioi_scores(
        logits,
        correct_names,
        incorrect_names
    )

    return {
        "mean_logit_difference": scores.mean(),
        "accuracy": (scores > 0).mean(),
        "individual_scores": scores
    }


clean_logits = model(tokens)

baseline = evaluate_ioi(clean_logits)


print("Baseline logit difference:",
      baseline["mean_logit_difference"])

print("Baseline accuracy:",
      baseline["accuracy"])
```

```python
# Cache Attention-Head Activations

# [batch, position, head, d_head]

# We have another edge here, individual attention head could be inspected
# rather than replacing entire attention layer.


# now you can access all the outputs of attention heads

# this is reference cache containing the original computation
# a clean experimental condition
clean_logits, clean_cache = model.run_with_cache(
    tokens,
    names_filter=lambda name: name.endswith("attn.hook_z")
)

for layer in range(model.cfg.n_layers):
    name = f"blocks.{layer}.attn.hook_z"
    activation = clean_cache[name]

    print(
        f"Layer {layer}:",
        tuple(activation.shape)
    )


```

```python
def ablate_head(z, hook, head_index):
    """
    Ablate one attention head's output at every position.
    z shape: [batch, position, head, d_head]
    """
    z = z.clone()
    z[:, :, head_index, :] = 0

    return z


def run_head_ablation(
    tokens,
    layer,
    head
):
    hook_name = f"blocks.{layer}.attn.hook_z"

    logits = model.run_with_hooks(
        tokens,
        fwd_hooks=[
            (
                hook_name,
                lambda z, hook: ablate_head(
                    z, hook, head
                )
            )
        ]
    )

    return evaluate_ioi(logits)
```

```python
# testing one head straightaway to see the effect of ablating it.
# please consider as intervention, to perform a behavior analysis.


test_layer = 9
test_head = 9

result = run_head_ablation(
    tokens,
    test_layer,
    test_head
)

print("Original score:",
      baseline["mean_logit_difference"])

print("Ablated score:",
      result["mean_logit_difference"])

print("Score change:",
      result["mean_logit_difference"]
      - baseline["mean_logit_difference"])
```

```python
# now let's play a big ablating
# ablating and sweeping all the heads of each layer

# what we are measuring is

# Layer index.

# Head index.

# Original logit difference.

# Ablated logit difference.

# Difference between the two.

# Accuracy after ablation.

```

```python
results = []

baseline_score = baseline["mean_logit_difference"]

for layer in range(model.cfg.n_layers):
    for head in range(model.cfg.n_heads):
        result = run_head_ablation(
            tokens,
            layer,
            head
        )
        ablated_score = result["mean_logit_difference"]
        results.append({
            "layer": layer,
            "head": head,
            "baseline_score": baseline_score,
            "ablated_score": ablated_score,
            "score_change": ablated_score - baseline_score,
            "accuracy": result["accuracy"]
        })

results_df = pd.DataFrame(results)

results_df.to_csv(
    "gpt2_head_ablation_results.csv",
    index=False
)

display(
    results_df.sort_values(
        "score_change"
    ).head(20)
)
```

### Understanding Score Changes in Head Ablation

When we ablate an attention head and evaluate the logit difference, we compute:
$$\text{Score Change} = \text{Ablated Logit Difference} - \text{Baseline Logit Difference}$$

* **Negative Score Change (Performance Drops):** These heads are **critical** for performing the IOI task. When silenced, the model struggles to identify the correct name over the incorrect name. These typically represent **Name Movers** (which write the correct name's representation to the output token) or **S-Inhibition / Duplicate Token Heads**.
* **Positive Score Change (Performance Improves):** These heads are **inhibitors** or **Negative Name Movers**. They normally write features that push the model toward outputting the incorrect name (the duplicate subject). Silencing them actually makes the model more confident in predicting the correct name.

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Pivot the results to create a 2D matrix (Layers x Heads)
ablation_matrix = results_df.pivot(index="layer", columns="head", values="score_change")

# Plot the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(
    ablation_matrix,
    cmap="RdBu",
    center=0.0,
    annot=False,
    cbar_kws={'label': 'Logit Difference Change'}
)
plt.title("Impact of Single-Head Ablation on IOI Logit Difference")
plt.xlabel("Attention Head Index")
plt.ylabel("Layer Index")
plt.show()
```

### Key Findings from Single-Head Ablation Sweep

We can find which heads are the most critical for solving the task by looking at the largest decreases in logit differences (negative score change). Run the cell below to print the top 10 most critical heads in the model.

```python
print("Top 10 most critical attention heads for the IOI task (largest negative change):")
display(results_df.sort_values("score_change").head(10))
```

### Separating Critical and Inhibitory Attention Heads

We can separate the heads based on their behavior upon ablation:
1. **Critical Heads (Negative Change):** Removing them harms performance because they actively help solve the task (e.g., Name Movers).
2. **Inhibitory Heads (Positive Change):** Removing them improves performance because they oppose the correct answer (e.g., Negative Name Movers).

```python
# 1. Critical Heads (Largest Negative Change)
critical_heads = results_df[results_df['score_change'] < 0].sort_values('score_change')

print("--- TOP 15 CRITICAL HEADS (Ablating harms the circuit) ---")
display(critical_heads.head(15))

# 2. Inhibitory Heads (Largest Positive Change)
inhibitory_heads = results_df[results_df['score_change'] > 0].sort_values('score_change', ascending=False)

print("\n--- TOP 15 INHIBITORY HEADS (Ablating improves the circuit) ---")
display(inhibitory_heads.head(15))
```
